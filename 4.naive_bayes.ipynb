```python
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
class naive_bayes:
    def __init__(self):
        self.priors = {}
        self.word_probability = {}
        self.classes = None

    def fit(self, X, y):
        self.classes = np.unique(y)

        for c in self.classes:

            x_mask = (y==c)
            pi_c = np.sum(x_mask)/len(y)

            x_c = X[x_mask]
            word_prob = np.sum(x_c, axis=0)
            word_prob +=1

            p_jc = word_prob/np.sum(word_prob)

            self.priors[c] =pi_c
            self.word_probability[c]= p_jc

    def predict(self, X):
        predictions = []

        for x in X:
            scores = []
            for c in self.classes:
                pi_c = self.priors[c]
                p_jc = self.word_probability[c]

                log_prior = np.log(pi_c)

                log_word = np.log(p_jc)
                before_sum = x* log_word

                log_word_prob = np.sum(before_sum)

                final_term = log_prior + log_word_prob

                scores.append(final_term)

            best_index = np.argmax(scores)
            prediction = self.classes[best_index]
            predictions.append(prediction)
            
        return np.array(predictions)        
```
